# Lesson 05 · Exercise 2 — Notebook top-to-bottom

Goal: load a CSV, display a dataframe, and plot a chart. This notebook must run top-to-bottom (`Kernel > Restart & Run All`).

## 1. Imports

We keep `%matplotlib inline` at the very top so every plot renders inline.

```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Load the CSV

```python
df = pd.read_csv('exercise2_data.csv')
df.head()
```

## 3. Dataframe display

Typing `df` (last expression) renders a rich HTML table, not `print(df)`.

```python
df.describe()
```

## 4. Plot the chart

```python
fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(df['date'], df['revenue'], label='revenue')
ax.bar(df['date'], df['users'], alpha=0.4, label='users', width=0.5)
ax.set_title('Revenue vs Users (90 days)')
ax.legend()
plt.show()
```

## 5. Conclusion

This notebook runs cleanly top-to-bottom: load -> display -> plot.
