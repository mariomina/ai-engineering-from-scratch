# Lesson 05 · Exercise 1 — `%timeit`

Compare **list comprehension** vs **NumPy** for creating an array of 100,000 random numbers.

```python
import numpy as np

size = 100_000
values = list(range(size))
```

## Micro-benchmark with `%timeit`

`%timeit` runs the code many times and reports the average (mean + std).

```python
%timeit [np.random.randn() for _ in range(size)]
```

```python
%timeit np.random.randn(size)
```

## Wall-clock with `%%time`

`%%time` runs the cell once and reports total elapsed time.

```python
%%time
import numpy as np
list_vals = [np.random.randn() for _ in range(size)]
numpy_vals = np.random.randn(size)
```
